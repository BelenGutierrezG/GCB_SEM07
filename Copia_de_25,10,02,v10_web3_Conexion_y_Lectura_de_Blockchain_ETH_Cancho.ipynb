**Funciones a través de `ethereum_tester`:**

* `mine_blocks(num_blocks=1)`: Mina un número específico de bloques en la blockchain de prueba.

```python
def get_block_details(web3_instance, block_number):
    """
    Retrieves and prints details of a specific block from a Web3 instance.

    Args:
        web3_instance: An instance of the Web3 class connected to a blockchain.
        block_number: The number of the block to retrieve.
    """
    try:
        bloque = web3_instance.eth.get_block(block_number)

        print(f"Detalles del bloque {block_number}:")
        print(f"Número de bloque: {bloque['number']}")
        print(f"Hash del padre: {bloque['parentHash'].hex()}")
        print(f"Hash del bloque: {bloque['hash'].hex()}")
        print(f"Transacciones incluidas: {len(bloque['transactions'])}")

        if bloque['transactions']:
            print("\nTransacciones:")
            for transaccion_hash in bloque['transactions']:
                transaccion = web3_instance.eth.get_transaction(transaccion_hash)
                print(f"  Transacción {transaccion_hash.hex()}:")
                print(f"    Remitente: {transaccion['from']}")
                print(f"    Destinatario: {transaccion['to']}")
                print(f"    Valor: {web3_instance.from_wei(transaccion['value'], 'ether')} Ether")
        else:
            print("No hay transacciones en este bloque.")

    except Exception as e:
        print(f"Error al obtener detalles del bloque {block_number}: {e}")
```
